```python
texts = [
    "Bonjour, je vous contacte pour confirmer notre rendez-vous de demain.",
    "Bonjour, veuillez trouver ci-joint le document demandé.",
    "Bonjour, pouvez-vous m'envoyer le rapport avant vendredi ?",
    "Merci pour votre message, je vous répondrai demain.",
    "Bonjour, la réunion est prévue à dix heures.",
    "Pouvez-vous confirmer votre présence à la réunion ?",
    "Bonjour, je vous envoie les informations demandées.",
    "Merci de me contacter lorsque vous serez disponible.",
    "Bonjour, le cours commence à neuf heures.",
    "Veuillez trouver ci-joint votre facture.",

    "Félicitations ! Vous avez gagné un cadeau gratuit. Cliquez ici pour le recevoir.",
    "Offre exceptionnelle ! Gagnez 1000 euros maintenant, cliquez sur ce lien.",
    "Vous avez gagné un prix incroyable ! Réclamez votre récompense maintenant.",
    "URGENT ! Cliquez ici pour recevoir votre cadeau gratuit.",
    "Félicitations, vous êtes le gagnant d'une somme de 5000 euros.",
    "Profitez de cette offre spéciale et gagnez de l'argent rapidement.",
    "Vous avez été sélectionné pour recevoir un cadeau gratuit.",
    "Cliquez immédiatement pour réclamer votre récompense.",
    "Gagnez de l'argent facilement depuis chez vous !",
    "Dernière chance ! Vous avez gagné un prix exceptionnel."
]

sh = [
    "ham", "ham", "ham", "ham", "ham",
    "ham", "ham", "ham", "ham", "ham",
    "spam", "spam", "spam", "spam", "spam",
    "spam", "spam", "spam", "spam", "spam"
]
```

```python
import string

texts_clean = []

for text in texts:
    text = text.lower()
    text = text.translate(str.maketrans('', '', string.punctuation))
    texts_clean.append(text)

print(texts_clean)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer()

X = vectorizer.fit_transform(texts_clean)

print(X.toarray())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    sh,
    test_size=0.2,
    random_state=42
)
```

```python
from sklearn.linear_model import LogisticRegression

model_lr = LogisticRegression(max_iter=1000)

model_lr.fit(X_train, y_train)
```

```python
from sklearn.svm import SVC

model_svm = SVC()

model_svm.fit(X_train, y_train)
```

```python
from sklearn.ensemble import RandomForestClassifier
model_rf = RandomForestClassifier(random_state=42)
model_rf.fit(X_train, y_train)
```

```python
from sklearn.model_selection import cross_val_score

scores_lr = cross_val_score(model_lr, X_train, y_train, cv=5)
scores_svm = cross_val_score(model_svm, X_train, y_train, cv=5)
scores_rf = cross_val_score(model_rf, X_train, y_train, cv=5)

print("Logistic Regression :", scores_lr)
print("SVM :", scores_svm)
print("Random Forest :", scores_rf)
```

```python
print("Moyenne Logistic Regression :", scores_lr.mean())
print("Moyenne SVM :", scores_svm.mean())
print("Moyenne Random Forest :", scores_rf.mean())
```

```python
from sklearn.metrics import accuracy_score

y_pred_lr = model_lr.predict(X_test)
y_pred_svm = model_svm.predict(X_test)
y_pred_rf = model_rf.predict(X_test)

print("Accuracy Logistic Regression :", accuracy_score(y_test, y_pred_lr))
print("Accuracy SVM :", accuracy_score(y_test, y_pred_svm))
print("Accuracy Random Forest :", accuracy_score(y_test, y_pred_rf))
```

Analyse des résultats

Les résultats montrent que la précision sur l'ensemble de test est de 100 % pour la régression logistique, et de 75 % pour le SVM et Random Forest.

La validation croisée donne une moyenne de 56,7 % pour les trois modèles. Les résultats doivent être interprétés avec prudence car le jeu de données utilisé est de petite taille.

```python
from sklearn.metrics import confusion_matrix, classification_report

print("Logistic Regression")
print(confusion_matrix(y_test, y_pred_lr))
print(classification_report(y_test, y_pred_lr))
```

```python
print("SVM")
print(confusion_matrix(y_test, y_pred_svm))
print(classification_report(y_test, y_pred_svm))
```

```python
print("Random Forest")
print(confusion_matrix(y_test, y_pred_rf))
print(classification_report(y_test, y_pred_rf))
```

Pour Logistic Regression, la matrice de confusion est [[2, 0], [0, 2]]. Cela signifie que les 2 messages Ham et les 2 messages Spam ont été correctement classés. Le modèle obtient donc une accuracy de 100 %.

Pour SVM, la matrice est [[2, 0], [1, 1]]. Les 2 messages Ham sont correctement classés, mais un message Spam est mal classé comme Ham. L'accuracy est donc de 75 %.

Pour Random Forest, on obtient le même résultat que pour SVM : 75 % d'accuracy.

La validation croisée donne une moyenne de 56,7 % pour les trois modèles. Comme notre dataset contient seulement 20 messages, ces résultats restent limités et doivent être interprétés avec prudence.
